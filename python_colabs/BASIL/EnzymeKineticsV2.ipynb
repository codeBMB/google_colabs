<u>A. Michaelis-Menten Kinetics</u>

A measured rate can be used to derive information about the amount of enzyme present in a reaction sample, or about the effects of different experimental conditions on rate. The Michaelis-Menten model for enzyme activity treats an enzyme catalyzed reaction as a simple two-step process; the first, association of substrate and enzyme to form enzyme-substrate complex, and the second, conversion of substrate to product:


$$E + S \rightleftharpoons ES \rightarrow E + P$$

The two-step nature of this mechanism is reflected in the dependence of rate on the concentrations of the two participants entering the reaction, substrate and enzyme.  At relatively low concentrations of substrate, the reaction displays overall second-order kinetics, with the rate dependent on concentrations of both substrate and enzyme.  In contrast, at saturating levels of substrate, the reaction rate becomes independent of substrate concentration – the reaction becomes first-order, dependent only on the concentration of enzyme.  Mathematically, the reaction rate at saturating substrate concentration is called $V_{max}$ with a rate constant $k_2$ (also called $k_{cat}$ (catalytic rate constant) or simply the turnover number):


$$V_{max}=k_{cat}[E]$$

Because $V_{max}$ depends only on the turnover number and the concentration of enzyme, we can use measured $V_{max}$ as an indication enzyme’s activity – that is, the amount of substrate molecules that will react, or “turn over,” in a given period of time in the presence of a sample of enzyme. In essence, this is the quantification of the qualitative enzyme activity assay that was performed in the previous BASIL module (Module 10: Protein Activity Assay) because the assay was carried out under saturating substrate conditions.


There is a host of factors that can affect measured enzyme activity, which can seem complicated at first glance but actually serve to make that measured activity a useful parameter. First, the activity is an extrinsic property of the enzyme. An enzyme’s activity depends on its concentration. The more enzyme, the more activity. Therefore, activity is a handy stand-in for amount of enzyme, and can be used to track the presence of the enzyme during purification.  Thus, experiments in which one of these parameters is varied provide valuable information about an enzyme’s optimal reaction conditions, ultimately providing insight into biological function.


In practice, Michaelis-Menten kinetic constants are determined experimentally by carrying out the reaction with a constant amount of enzyme under varying substrate concentrations. The data are plotted and the slope reflects the rate of the reaction (velocity) at that substrate concentration. The velocity rates are then plotted versus substrate concentration to generate the Michaelis-Menten plot (Figure 2). More information on these ideas is provided below.

<u>B. Determining optimum assay conditions</u>

Before obtaining kinetic measurements the reaction conditions should be optimized for the specific enzyme. Each enzyme will have a unique temperature, pH, salt concentration and buffer composition preference. These optimal conditions can vary widely but data about organism homology and enzyme mechanism can provide a starting point in their determination. For example, enzymes that originate in thermophiles likely function best at high temperatures while hydrolases originally functioning in the stomach might function best at low pH values. Since each enzyme is unique, kinetic experiments must be designed methodically based on previous bioinformatic conclusions or experimental observations.

<u>C. Determination of enzyme specific kinetic parameters</u>`

The previous BASIL module (Module 10: Protein Activity Assay) focused on an enzyme assay to determine the presence of an enzymatic reaction using standard conditions. The next set of experiments involve characterizing the substrate affinity and the reaction rate of an enzyme. When determining $K_M$ and $V_{max}$ values the experimental design has to be carefully tailored.

If one is determining these steady-state kinetic parameters, substrate concentrations must vary from assay to assay. A series of assays is performed so that the concentration of substrate giving $V_{max}/2$ (the $K_M$ value) lies in the middle of the series. All substrate concentrations should be higher than enzyme concentration so that the steady-state assumption holds, but the concentrations when plotted must bracket the KM value. Depending on factors such as stability of the substrate or order of addition of reagents, assay directions may have to be altered slightly or greatly.

As in the theoretical example shown, investigators weigh the acceptability of the absorbance curve (Figure 3). The experiment nearly always has to be repeated to optimize the relative amounts of substrate and enzyme. If the curve rises within seconds and then plateaus, too much enzyme was present and needs to be diluted; if, on the other hand, the $A_{405}$ curve does not increase significantly over a reasonable time period more enzyme needs to be added (or the enzyme simply has little to no activity on the substrate as tested). Once conditions are optimized, a range of time over which the $A_{405}$ value increases linearly is identifiable. The first few minutes of the reaction shown in Figure 3 display a linear rise.  The slope of this line is the rate (or velocity) of the enzyme-catalyzed reaction.

<u>D. Beer-Lambert Law. </u>

Velocity is the change in product concentration over the change in time (Δ\[P\]/Δt). Absorbance data must therefore be converted to concentration of product. This is accomplished using the  Beer-Lambert Law:

$$Equation 1: A = ecl$$

$$Absorbance = coefficient * concentration * pathlength(1 cm)$$

The extinction coefficient (e) is a number specific to the product being studied at a specific wavelength. For example , PNP measured at 405 nm has an extinction coefficient of $18,320 cm^{-1} M^{-1}[1]$. Each absorbance measurement is converted to the concentration of product and plotted in a graph of product concentration versus time (Figure 4). The slope of the graph shows the change in product concentration over the change in time (Δ\[P\]/Δt) or the velocity of the enzyme-catalyzed reaction. This visualizes how much of the substrate is being converted to product over time. For example, if the linear portion of the progress curve has a slope Δ\[A\]/Δt \= 0.100/min and if the cuvette pathlength is 1.0 cm (typical), then dividing Δ\[A\]/Δt by extinction coefficient gives $0.100/min/(18,320 cm^{-1} M^{-1} x 1.0 cm) = 5.5 * 10^{-6} M/min$ for the velocity, V.

*C. The assay.*

1. Following your experimental design table, add buffer and substrate to the cuvette. *Hint: add buffer first so that the acetonitrile in the substrate solution does not etch the plastic of the cuvette.*  
2. Place the cuvette in the spectrophotometer and zero the instrument.  
3. Add enzyme (either 10 uL of 10 mg/ml chymotrypsin or an appropriate volume of the protein of interest). Mix quickly but thoroughly, either by aspirating up and down a few times with the pipet or by inversion.   
4. Follow the absorbance at 405 nm over time.   
   * Take measurements every 30 seconds or set the instrument to do this automatically.  
   * Follow the reaction. This may take up to 30 minutes. As the reaction proceeds, you will likely observe one of three things:  
      * If the absorbance remains near zero and does not increase over the time of measurement, your enzyme has no (or very little) activity relative to your candidate substrate.  
      * If the absorbance rises gradually and in a linear manner over the course of your measurement, you are in the “initial rate” ($V_0$) range of your assay, which is ideal for kinetic measurements.  In this case, the slope of the absorbance versus time over the entire course of your reaction can be taken as measured rate.  
      * If the absorbance rises quickly but then flattens out (figure 3), your reaction has consumed available substrate.  This is fine as long as you have enough data points from the early minutes of your reaction to obtain a reliable linear phase of your curve.  In this case, the rate should be calculated as the slope of the line fit to the linear portion of the curve.  However, if the initial rise is so fast (less than a few minutes), you should repeat your assay with less enzyme.

*D. Data Analysis.*

1. Using the Beer-Lambert law, convert the absorbance readings to concentration of product.  
2. Plot \[P\] versus time for each substrate concentration experiment.  
3. Determine the rate as the slope of each line over the linear portion of each curve.    
4. Plot the data as a Michaelis-Menten or Lineweaver-Burk plot to determine $K_M$ and $V_{max}$ (see “Interpreting Results”, below).

#**Interpreting Results**


---



Plotting your data:  
Recall that a typical Michaelis-Menten plot includes enzyme velocities on the y-axis and substrate concentrations on the x-axis (Figure 2). Use this kind of plot. It can be very satisfying to keep track of your plot as you collect more points\! It can also tell you whether you need to go higher or lower in substrate concentration. If, while you are plotting your V vs \[S\] points,  you do not see any saturation plateau, you are not bracketing  the KM value and you need to add more substrate. If you don’t see any increase in V with increasing \[S\], you may already be at Vmax conditions and you need to add less substrate.




When you do have data that display a hyperbola, try to fit your data to the equation of a hyperbola. The equation is $V = (M1 * [S])/(M2+[S])$; where M1 is the $V_{max}$ value and M2 is the $K_M$ value. If you have a good graphing calculator, you may be able to do this with your calculator. If you have a graphing program such as Kaleidagraph (Synergy) you can ask your graphing program to determine these constants. If you do not have these resources, try the online data fitting program at   
[https://mycurvefit.com](https://mycurvefit.com). In that program, choose the fitting method “non-linear” “Michaelis-Menten”. Enter your data into the small spreadsheet window below the plot, and click “export”. The program will enter your data, plot it, and give you a $V_{max}$ value and a $K_M$ value.

If you are in the enviable position of testing various substrates to determine which one is the better one, and perhaps the more likely natural one, you can utilize the $V_{max}/K_M$ value. Simply divide the $V_{max}$ by the $K_M$ to obtain this parameter. Relative Vmax/KM values are regarded as indicators of substrate specificity. It is often true (though not always, *e. g.,* for highly regulated enzymes) that the higher the Vmax/KM value is, the better the substrate is.



Keep in mind that the $K_M$ and $V_{max}$ measurements are dependent upon the environmental conditions of the assay, the identity of the substrate, and the amount of enzyme present. If they are determined under different conditions and compared, that might provide insight into the enzyme function. For example, a lower $K_M$ at a lower pH could indicate that the protonation state of the substrate and/or enzyme are important in the activity. It may also indicate that the enzyme evolved to be most efficient at lower pHs. A comparison of the kinetic data with different substrates can indicate substrate preference of the enzyme. A comparison of the kinetic data in the absence and presence of a ligand can indicate if the ligand acts as an inhibitor.

**Note and caution:** If you do not have access to the internet, it is possible to do a Lineweaver-Burk plot (Figure 6\) using graph paper and pencil. For this you must convert your data to their reciprocal values, and plot them instead of the direct data. If you take the reciprocal of both sides of the Michaelis-Menten equation, you arrive at the Lineweaver-Burk equation, which defines a line rather than a curve.

Lineweaver-Burk equation :	$$1/v = 1/Vmax + (K_M/V_{max})(1/[S])$$

This is easier to plot and easier to describe via slope and intercepts, than the plot of a hyperbola.   
The y-axis is now 1/v and the x-axis is 1/\[S\].  Compare this to the equation of a line and it is clear that the slope of a Lineweaver-Burk plot is KM/Vmax , the y-intercept is 1/Vmax.  The x-intercept is \-1/KM.


<center> <img src="https://raw.githubusercontent.com/codeBMB/google_colabs/BASIL/python_colabs/BASIL/images/LineweaverBurkPlot.png" width="500">

**Figure 6: A Lineweaver-Burk plot of kinetic data: 1/v vs 1/[S].**


You can find the kinetic values of KM and Vmax  using a ruler and a pencil on actual paper graph, and it seems so easy as to be the preferred method of determining Michaelis-Menten parameters. It should however, be the method of last resort, used only until one can do the fitting of a rectangular hyperbola. The reason for this is that a Lineweaver-Burk plot necessarily weights the very worst of your data. The lowest velocity points, which have the highest noise-to-signal contamination, lie furthest away from the origin, and will influence how you draw your line greatly. How you anchor your Lineweaver-Burk line creates huge uncertainty into where the y-intercept actually is. If you then use your error-vulnerable y-intercept-determined $1/V_{max}$ to solve for $K_M$ in the slope, both have been harmed by using the linear double-reciprocal Lineweaver-Burk plot.

# Michaelis-Menten Plot

The following code cell creates a fitted plot using the data input. Through this $K_m$ and $V_{max}$ values can be determined.

<br>The graph being plot sould look like the graph represented in Figure 2. If the graph looks incorrect, make sure the correct data is being put in for the *x_coords* and *y_coords*.

````python
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
import numpy as np

# Define the Michaelis-Menten equation
def michaelis_menten(S, Vmax, Km):
    return (Vmax * S) / (Km + S)

# Convert lists to numpy arrays for curve_fit
Substrate_Concentration_arr = np.array(x_coords)
Velocity_arr = np.array(y_coords)

# Use curve_fit to find the optimal parameters (Vmax, Km)
params, covariance = curve_fit(
    michaelis_menten,
    Substrate_Concentration_arr,
    Velocity_arr)
Vmax_fit, Km_fit = params

# Calculate R-squared for Michaelis-Menten fit
predictions_mm = michaelis_menten(Substrate_Concentration_arr, Vmax_fit, Km_fit)
ss_res = np.sum((Velocity_arr - predictions_mm) ** 2)
ss_tot = np.sum((Velocity_arr - np.mean(Velocity_arr)) ** 2)
R2_mm = 1 - (ss_res / ss_tot)

# Generate points for the fitted curve
S_fit = np.linspace(min(x_coords), max(x_coords), 100)
V_fit = michaelis_menten(S_fit, Vmax_fit, Km_fit)

# Plot the experimental data and the fitted curve
print(f"Fitted Michaelis-Menten Equation: V = ({Vmax_fit:.2f} * S) / ({Km_fit:.2f} + S)")
print(f"Fitted Vmax = {Vmax_fit:.2f} ")
print(f"Fitted Km =   {Km_fit:.2f} ")
print(f"R² =          {R2_mm:.2f}")

plt.figure(figsize=(10, 6))
plt.scatter(x_coords, y_coords, color='red', label='Experimental Data')
plt.plot(S_fit, V_fit, color='blue', label='Michaelis-Menten Fit')

plt.title('Michaelis-Menten Kinetics: Experimental Data vs. Fitted Curve')
plt.xlabel('Substrate Concentration [S]')
plt.ylabel('Velocity [V]')
plt.grid(True)
plt.legend()
plt.show()
````

# Lineweaver-Burk Plot

The Lineweaver-Burk Plot uses the recipricals of $V$ and $[S]$ values as the *x* and *y* axis. Figure 6 above shows the Lineweaver burk plot aswell as explaining its equation:

$$1/v = 1/Vmax + (K_M/V_{max})(1/[S])$$

<br>The code cell below will use the data input to create this plot. The *x* and *y* intercepts can be used to determine the $K_m$ and $V_{max}$ vlues using this plot.


````python
from scipy.stats import linregress

# Filter out zero-activity/concentration points and calculate reciprocals
valid_data = [(x, y) for x, y in zip(x_coords, y_coords) if x != 0.0 and y != 0.0]
reciprocal_S = np.array([1/x for x, _ in valid_data])
reciprocal_V = np.array([1/y for _, y in valid_data])

# Linear regression on 1/[S] vs 1/v
slope_lb, intercept_lb, r_value_lb, p_value_lb, std_err_lb = linregress(
    reciprocal_S, reciprocal_V
)
R2_lb = r_value_lb**2

# Calculate Km and Vmax from LB parameters
Vmax_lb = 1 / intercept_lb
Km_lb   = slope_lb * Vmax_lb

# Prints out results
print(f"--- Lineweaver-Burk Results ---")
print(f"Slope:      {slope_lb:.2f}")
print(f"Intercept:  {intercept_lb:.2f}")
print(f"R²:         {R2_lb:.2f}")
print(f"Vmax:       {Vmax_lb:.2f}")
print(f"Km:         {Km_lb:.2f}")


# Generate fit line extending into negative x-axis to show x-intercept (-1/Km)
x_min_fit = min(reciprocal_S.min(), -1/Km_lb) * 1.3
x_max_fit = reciprocal_S.max() * 1.1
S_fit_lb  = np.linspace(x_min_fit, x_max_fit, 200)
V_fit_lb  = slope_lb * S_fit_lb + intercept_lb

# Create the plot
plt.figure(figsize=(10, 6))
plt.scatter(reciprocal_S, reciprocal_V, color='blue', s=80,
            label='Experimental Data', zorder=5)
plt.plot(S_fit_lb, V_fit_lb, color='green', linestyle='--',
         label=f'Linear Fit: y = {slope_lb:.4f}x + {intercept_lb:.4f}\nR² = {R2_lb:.4f}')

# Creates the origin
plt.axhline(y=0, color='black', linewidth=0.8)
plt.axvline(x=0, color='black', linewidth=0.8)

# Marks the X and Y intercepts.
plt.scatter([-1/Km_lb], [0], color='red', s=100, zorder=6,
            label=f'X-intercept = -1/Km = {-1/Km_lb:.4f}')
plt.scatter([0], [1/Vmax_lb], color='orange', s=100, zorder=6,
            label=f'Y-intercept = 1/Vmax = {1/Vmax_lb:.4f}')

# Creates the labels for the plot.
plt.xlabel('1/[S]')
plt.ylabel('1/V')
plt.title('Lineweaver-Burk Plot')
plt.grid(True, alpha=0.3)
plt.legend(loc='upper left')
plt.show()
````

# Eadie-Hofstee Plot

````python
eh_S = [x for i, x in enumerate(x_coords) if y_coords[i] != 0.0 and x != 0.0]
eh_V = [y for i, y in enumerate(y_coords) if y != 0.0 and x_coords[i] != 0.0]

# Calculate v/[S] for the x-axis
v_over_S = np.array([v / s for v, s in zip(eh_V, eh_S)])
v_array  = np.array(eh_V)

# Linear regression: v = Vmax - Km * (v/[S])
# Y = intercept + slope * X → slope = -Km, intercept = Vmax
slope_eh, intercept_eh, r_value_eh, p_value_eh, std_err_eh = linregress(
    v_over_S, v_array
)
R2_eh   = r_value_eh**2
Vmax_eh = intercept_eh
Km_eh   = -slope_eh

print(f"--- Eadie-Hofstee Results ---")
print(f"Slope:      {slope_eh:.2f}   (= -Km)")
print(f"Intercept:  {intercept_eh:.2f}   (= Vmax)")
print(f"R²:         {R2_eh:.2f}")
print(f"Vmax:       {Vmax_eh:.2f}")
print(f"Km:         {Km_eh:.2f}")
print()

# Generate fit line from 0 to x-intercept
x_intercept_eh = Vmax_eh / Km_eh if Km_eh != 0 else np.max(v_over_S) * 1.5
x_max_plot = max(np.max(v_over_S), x_intercept_eh) * 1.1
X_fit_eh = np.linspace(0, x_max_plot, 200)
Y_fit_eh = slope_eh * X_fit_eh + intercept_eh

# Create the plot
plt.figure(figsize=(10, 6))
plt.scatter(v_over_S, v_array, color='darkorange', s=80,
            label='Experimental Data', zorder=5)
plt.plot(X_fit_eh, Y_fit_eh, color='blue', linestyle='--',
         label=f'v = {intercept_eh:.4f} + ({slope_eh:.4f})·(v/[S])\nR² = {R2_eh:.4f}')

# Mark y-intercept (Vmax) and x-intercept (Vmax/Km)
plt.scatter([0], [Vmax_eh], color='red', s=100, zorder=6,
            label=f'Y-intercept = Vmax = {Vmax_eh:.4f}')
plt.scatter([x_intercept_eh], [0], color='purple', s=100, zorder=6,
            label=f'X-intercept = Vmax/Km = {x_intercept_eh:.4f}')

plt.axhline(y=0, color='black', linewidth=0.8)
plt.axvline(x=0, color='black', linewidth=0.8)
plt.ylim(ymin=0)
plt.xlim(xmin=0)
plt.xlabel(f'v/[] ')
plt.ylabel(f'v ()')
plt.title('Eadie-Hofstee Plot')
plt.grid(True, alpha=0.3)
plt.legend(loc='upper right')
plt.show()
````
