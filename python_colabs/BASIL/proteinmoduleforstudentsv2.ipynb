# **Introduction**
---
Biochemical experiments require accurate determination of protein concentration, for example, for determining purification yields or to prepare a sample of appropriate concentration for analysis.  Protein quantitation usually relies on absorbance measurements. These measurements can be grouped into two broad categories: intrinsic absorbance and colorimetric/dye-binding (Table 1). **Intrinsic absorbance** measures absorbance of a protein without addition of any other reagents.  Most proteins absorb UV light with maximum absorbance at 280 nm due to absorbance of aromatic amino acid side chains (mainly tryptophan) and disulfide bonds; hence this assay is sensitive to the amino acid composition of the proteins (1, 2).  **Colorimetric and dye binding** assays are based on the change in absorbance of metal ions or organic dyes upon binding to proteins.  These methods are less sensitive to amino acid composition than intrinsic absorbance. When choosing an appropriate method, one must carefully consider potential interfering compounds.

Factors to consider in choosing an assay include: sensitivity and working/linear range; interfering substances; effect on the protein (*e.g.* $A_280$ does not destroy protein, but other methods do); time and cost constraints; and availability of protein standard (solution of known concentration).  Table 1 compares several methods; additional information can be found in the original references or in the product information for commercially available reagents.  

# **Beer-Lambert’s Law**
---
When we use spectrophotometry to determine concentration, we relate the amount of light absorbed by the substance to its concentration using the Beer-Lambert law (also known as Beer’s law):

$A=εcl$

where $A$ is absorbance, $c$ is the concentration of the solution, $l$ is the pathlength (usually 1 cm), and $ε$ is the extinction coefficient. Absorptivity is a number specific to the wavelength at which absorbance is measured and the substance in solution that absorbs light. Extinction coefficient, also

Because absorbance, A, has no units, the units of **ε** must cancel those of ***c*** and ***l,*** and consequently, must be in inverse concentration and inverse length, M<sup>-1</sup>· cm<sup>-1</sup>.  The pathlength term *l* often drops out because standard UV-vis cuvettes have a 1 cm pathlength.  The concentration term *c* is normally in units of molarity, but occasionally a given extinction coefficient may have non-SI units such as micromolarity (μM), or mass concentration such as mg/ml.  Note two confusing pairs of terms: **absorptivity** should not to be confused with absorbance, and **pathlength** (measured in cm) should not be confused with wavelength (measured in nm).  

For a pure protein, one could simply determine the concentration using the Beer-Lambert Law and the extinction coefficient for the protein if known.  An estimated value can be calculated using the protein’s amino acid sequence (3); a useful online resource to perform this calculation can be found at [https://web.expasy.org/protparam/](https://web.expasy.org/protparam/).   If the sequence of a protein is not available, a very rough rule of thumb can be applied:  a 1 mg/ml solution of protein of typical amino acid composition in a 1 cm cuvette would be expected to have an absorbance of about 1\.  That is, most proteins have a mass-based extinction coefficient of around $1 (mg/ml)^{-1}cm^{-1}$; while inexact, this estimate can be helpful during the purification of a protein.  

While convenient, intrinsic absorbance measurement has the disadvantage that many other substances in a protein-containing sample also absorb light in the ultraviolet range.  In particular, nucleic acids (both RNA and DNA) absorb significant amounts of light at 280 nm, and are often present in a cell extract.  For this reason, it is often preferable to measure protein concentration using a colorimetric assay, where one first adds substance which reacts specifically with protein with a resulting color change, followed by measuring the absorbance change due to the reaction.  Such colorimetric assays react with proteins by mass, not mole, and in a way that is (at least approximately) independent of sequence, so that one does not need to know the molar mass or amino acid composition to measure concentration.  Although there are several well-characterized colorimetric assays available (Table 1), this experiment uses the dye-based Bradford assay due to its ease of use and reliability.

#**Bradford Assay**
---
The Bradford Assay is a simple one-step method that can be performed as a standard assay (3.1 mL assay volume, requires 0.1 mL of protein, and a working range of approximately 0.1-1.0 mg/mL), in 96-well plates (same working range, but only 5 μL of protein is required) or as a micro assay (2 mL assay volume, requires 1 mL of dilute protein, 1-10 μg/mL working range).  The dye used in the Bradford assay is Coomassie Blue G-250, the same molecule commonly used for detecting the presence of proteins in polyacrylamide following electrophoresis.  In this assay, dye is mixed with the protein under acidic conditions.  The protein interacts with the dye *via* ionic and hydrophobic interactions of the polypeptide backbone and amino acid side chains.  (4)  The interaction stabilizes the anionic form of the dye, causing a color change from brownish-red to blue and a corresponding shift in the λ max from 465 nm (free dye) to 595 nm (dye-protein complex).  The reaction can occur in as little as 5 minutes.

#**Use of a standard curve**
---
When you use spectrophotometric absorbance, whether intrinsic or colorimetric, to determine concentration, you are assuming a linear relationship between concentration and absorbance.  Linearity should always be confirmed by using a standard curve and graphical analysis, particularly when using dye-based concentration assays. A standard curve is generated by measuring the absorbance of a protein standard at a variety of known concentrations.  The data are plotted with concentration (independent variable) on the x-axis, and absorbance (dependent variable) on the y-axis.  

If the standard protein behaves similarly to the unknown protein(s), and both are handled in the assay the same way, the inferred concentration should be reasonably accurate, though it should be noted that dye-binding assays respond differently to different protein standards.  When designing an experiment, you should carefully consider the choice of standard protein and the choice of the assay.  The standard protein of choice is often bovine serum albumin (BSA).  It is small, soluble, well-behaved and relatively inexpensive. Other issues to consider when choosing an assay include: time required, sensitivity, difficulty, cost, and unfavorable interactions with components in the solution (detergents, reducing agents, etc.).

**Procedure** (Based on product information from Sigma-Aldrich (5). )

<u>A.  Plan your experiment</u>

1. Prepare a table of standards and samples, either in a spreadsheet on your computer, or on paper in your lab notebook.  Cuvettes should include:  
   * One (two, for a double-beam instrument) **zero**, containing 3 mL of water, without dye or protein, for zeroing the instrument.

   * One for each unknown sample.  
2. Assemble the required number of cuvettes in a sample holder and label them with permanent marker in an area which is not in the way of the light path.

<u>A. Preparing BSA Standards</u>

Using the buffer that your protein is stored in, prepare a series of 5-7 solutions of bovine serum albumin (BSA).  In a standard cuvette-based assay, standard solutions should range from 0.10 to 1.00 mg/mL.       

<u>B. Bradford Assay</u>

Bradford Assay in 3.10 mL volume (*for 1.03 mL in parentheses*)

1. If Bradford reagent was obtained as a concentrated stock, dilute as instructed.  
2. Add 3 mL (*1 mL*) of Bradford Reagent to each cuvette except for the “zero” cuvette(s), to which 3 mL (*1 mL*)  of water is added.  Add 100 μL (*30 μL*) of water to the blank, and 100 μL (*30 μL*)  of each BSA standard and unknowns to the rest.  
3. Cover your cuvettes with small squares of parafilm and mix each one several times by gentle inversion.  
4. Maintain at room temperature for 5-45 minutes.  
5. Record absorbances at 595 nm of a spectrophotometer.

Bradford 96-well Plate Assay (requires plate reader)

1. In each well add 5.00 μL of sample.  
2. Add 250 μl of Bradford Reagent to each well.  
3. Mix gently by shaking for 30 seconds.  
4. Maintain at room temperature for 5-45 minutes.  
5. Measure absorbance at 595 nm with the plate reader.

**Clean-up:**

Cuvettes should not be reused, as it is impossible to completely remove traces of reagent and protein from the plastic surface.

The formula you obtain from the graph of your own data can be used to calculate your unknown concentrations.

In the case of unknowns whose absorbance exceeds the linear portion of the curve (such as unknown 1 in the example), you should repeat the measurement using a diluted sample.  For example, you might decide to dilute an unknown by, first, combining 100 μL of sample with 400 μL of water.  This will produce a solution with one-fifth the concentration, which you can figure out using the $C_1V_1=C_2V_2$ equation; in this case $V_1 = 100 μL$ and $V_2 = 100 μL + 400 μL = 500 μL$.  You do not need to repeat the standard curve, as it remains valid, but you should take care to replicate the conditions you used previous, including incubation time between combining reagents and taking the absorbance readings.  Once the concentration of the diluted sample ($C_2$) is found, you can backtrack to find the concentration of the initial, more concentrated sample ($C_1$).

If your protein is relatively pure, for example, after isolation by affinity chromatography, and you know its molar mass, you can convert the concentration you measured from mass/volume to molar.  For example, if you know your protein’s molar mass to be 43,000 and it has a concentration of 0.46 mg/mL, you can calculate its molar (or micromolar) concentration as follows:

$Protein Conc= \frac{0.46 mg}{mL}*\frac{1000 mL}{L}*\frac{1 g}{1000 mg}*\frac{1 mol protein}{43,000 g} = 11 μM$

#  **Data Entry**

In the following cell, your collected data can be entered in the x_data and y_data sections. Data should be entered separated by commas. You should only enter data for the x_data and y_data_raw lines.

```python
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
import numpy as np

def linear_function(x, slope, intercept):
    return slope * x + intercept

# ENTER YOUR DATA HERE
x_data = [0,0.2,0.4,0.6,0.8,1] # Initial Protien (mg/ml)
y_data_raw = [.449,.635,.805,.97,1.09,1.213] # Abs. (595 nm)

# Perform baseline correction by subtracting the blank's absorbance
baseline_value = y_data_raw[0] # Assuming the first point is the blank
y_data = [y - baseline_value for y in y_data_raw]

params, covariance = curve_fit(linear_function, x_data, y_data)
slope, intercept = params

# Calculate standard errors
perr = np.sqrt(np.diag(covariance))
slope_error = perr[0]
intercept_error = perr[1]

# Prints the error values
print(f"Fitted Linear Equation: y = {slope:.4f}x + {intercept:.4f}")
print(f"Slope: {slope:.4f} +/- {slope_error:.4f}")
print(f"Intercept: {intercept:.4f} +/- {intercept_error:.4f}")

# This generates the plot for your raw data.
plt.scatter(x_data, y_data, label= 'Data')
plt.plot()

# This generates the linear least square fit line.
x_trendline = np.linspace(np.min(x_data), np.max(x_data), 100)
y_trendline = linear_function(x_trendline, slope, intercept)
plt.plot(x_trendline, y_trendline, color='red', linestyle='--', label=f'Trendline: y = {slope:.4f}x + {intercept:.4f}')

# This generates the labels.
plt.xlabel("Initial Protien (mg/ml)")
plt.ylabel("Abs. (595 nm)")
plt.title("Absorbance vs. Initial Protien Concentration")

# This generates the legend that includes the trendline.
plt.legend()
plt.show()
```
